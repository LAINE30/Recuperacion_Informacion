```python
import gensim.downloader as api

model = api.load("word2vec-google-news-300")
```

```python
model 
```

```python
vec = lambda word: model[word]
```

```python
len(model['dog'])
```

```python
vec("word")
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cosine_similarity([vec('cat')], [vec('dog')])
```

```python
cosine_similarity([vec('king')], [vec('queen')])
```

```python
vec_man = vec("man")

vec_woman = vec("woman")

dist = vec_woman -vec_man
```

```python
vec_king = vec("king")
x = vec_king + dist
```

```python
cosine_similarity([x], [vec('queen')])
```

```python
model.most_similar(x)
```

```python
from sklearn.descomposition import PCA
pca = PCA(components=2)
```

```python
vman = pca.fit_transform([vec_man])
vwoman = pca.fit_transform([vec_woman])
vking = pca.fit_transform([vec_king])
vqueen = pca.fit_transform([vec('queen')])
vx = pca.fit_transform([x])
```

```python
plt.figure(figsize=(10,10))
plt.scatter(vman[0][0], vman[0][1], color='blue')
plt.scatter(vwoman[0][0], vwoman[0][1], color='red')
plt.scatter(vking[0][0], vking[0][1], color='green')
plt.scatter(vqueen[0][0], vqueen[0][1], color='orange')
plt.scatter(vx[0][0], vx[0][1], color='purple')
```
